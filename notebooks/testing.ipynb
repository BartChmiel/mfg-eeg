```python
import numpy as np

data = np.load("../out/mfg_Fp1_Fp2.npz")
print(list(data.keys()))
```

```python
print("GC shape:", data["gc"].shape)
print("Scores shape:", data["scores"].shape)
print("Pierwsze 10 GC:", np.round(data["gc"][:10], 4))
print("Min/Max GC:", np.nanmin(data["gc"]), np.nanmax(data["gc"]))
```

```python
import matplotlib.pyplot as plt

ms = data["ms"]
gc = data["gc"]
plt.plot(ms, gc)
plt.xlabel("lag [ms]")
plt.ylabel("GC(delta)")
plt.title("Fp1 → Fp2")
plt.axvline(0, color="k", lw=0.5)
plt.tight_layout()
plt.show()
```
